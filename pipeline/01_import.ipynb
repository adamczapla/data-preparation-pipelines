# Import des Datensatzes

Die Rohdaten liegen als Excel-Arbeitsmappe mit zwei Tabellenblättern vor. Beide Tabellenblätter werden eingelesen.

```python
from src.paths import RAW_DATA_DIR, IMPORTED_DATA_DIR
import pandas as pd

workbook = pd.ExcelFile(RAW_DATA_DIR / "online_retail_II.xlsx")

sheets = [
    workbook.parse(sheet_name=sheet) for sheet in workbook.sheet_names
]
```

## Zusammenführen der Tabellenblätter

Im Data Profiling wurde festgestellt, dass beide Tabellenblätter dieselben Spalten in identischer Reihenfolge sowie übereinstimmende Datentypen besitzen.
Daher werden sie zu einem gemeinsamen Datensatz zusammengeführt.

```python
dataset = pd.concat(sheets, ignore_index=True)
```

## Konvertierung der Datentypen

Im Data Profiling wurde festgestellt, dass die **Spalten** **Invoice**, **StockCode** und **Description** den Datentyp `object` besitzen.
Da diese **Spalten** ausschließlich Identifikatoren beziehungsweise Textinformationen enthalten, werden sie in den pandas-Datentyp `string` konvertiert.

Die Konvertierung ist aus zwei Gründen erforderlich. Zum einen können gemischte `object`-Datentypen nicht zuverlässig im `Parquet`-Format gespeichert werden.
Zum anderen vereinfacht ein einheitlicher Datentyp die weitere Verarbeitung, da Stringoperationen ohne vorherige Typkonvertierung auf alle Werte angewendet
werden können.

```python
dataset["Invoice"] = dataset["Invoice"].astype("string")
dataset["StockCode"] = dataset["StockCode"].astype("string")
dataset["Description"] = dataset["Description"].astype("string")
```

## Speichern des importierten Datensatzes

Der zusammengeführte und typisierte Datensatz wird im `Parquet`-Format gespeichert. Die erzeugte Datei dient als Eingabe für das Cleaning.

```python
dataset.to_parquet(IMPORTED_DATA_DIR / "online_retail_II.parquet")
```
